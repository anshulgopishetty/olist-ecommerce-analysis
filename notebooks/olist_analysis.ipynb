```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns 

## CREATING CUSTOMERS DATAFRAME:

customers_df = pd.read_csv(
    '../data/olist_customers_dataset.csv'
)

customers_df.info()
print(customers_df.shape)
print(customers_df.dtypes)
```

```python
## CREATING ORDERS DATAFRAME:

orders_df = pd.read_csv(
    "../data/olist_orders_dataset.csv"
)

orders_df.info()
print(orders_df.shape)
print(orders_df.dtypes)
orders_df.isna().sum()
```

```python
## CREATING ORDER_ITEMS DATAFRAME:

order_items_df = pd.read_csv(
    "../data/olist_order_items_dataset.csv"
)

order_items_df.info()
print(order_items_df.shape)
print(order_items_df.dtypes)
```

```python
## CREATING PRODUCTS DATAFRAME:

products_df = pd.read_csv(
    '../data/olist_products_dataset.csv'
)

products_df.info()
print(products_df.shape)
print(products_df.columns)
```

```python
## CREATING ORDER_REVIEWS DATAFRAME:

order_reviews_df = pd.read_csv(
    '../data/olist_order_reviews_dataset.csv'
)

order_reviews_df.info()
print(order_reviews_df.shape)
print(order_reviews_df.columns)
```

```python
## CREATING ORDER_PAYMENTS DATAFRAME:

order_payments_df = pd.read_csv(
    '../data/olist_order_payments_dataset.csv'
)

order_payments_df.info()
print(order_payments_df.shape)
print(order_payments_df.columns)


```

```python
## CREATING SELLERS DATAFRAME:

sellers_df =  pd.read_csv(
    '../data/olist_sellers_dataset.csv'
)

sellers_df.info()
print(sellers_df.shape)
print(sellers_df.columns)

```

```python
#Finding Total Orders For Each Order Status:

orders_df.groupby('order_status').agg(
    total_orders = ('order_id', 'count')   
)
```

```python
## Counting Missing Values For Each Order Status By Grouping It:

temp_df = (orders_df.assign(
    missing_approved = orders_df['order_approved_at'].isna(),
    missing_carrier = orders_df['order_delivered_carrier_date'].isna(),
    missing_customer_deliver_date = orders_df['order_delivered_customer_date'].isna()
)
.groupby('order_status')[['missing_approved','missing_carrier','missing_customer_deliver_date']]
.sum()
) 

temp_df
```

```python
## Missing "Delivered" and "order_deliverd_customer_date" Values Details:

orders_df[(
    orders_df['order_status'] == "delivered") & (
        orders_df['order_delivered_customer_date'].isna()
    )
][['order_id','order_purchase_timestamp','order_estimated_delivery_date']]
```

```python
## Converting columns(currently with object dtype) with dates to DateTime Format:

date_cols_orders = ['order_purchase_timestamp','order_approved_at','order_delivered_carrier_date','order_delivered_customer_date','order_estimated_delivery_date']

orders_df[date_cols_orders] = orders_df[date_cols_orders].apply(pd.to_datetime)

order_items_df['shipping_limit_date'] = pd.to_datetime(
    order_items_df['shipping_limit_date']
)

date_cols_reviews = ['review_creation_date','review_answer_timestamp']

order_reviews_df[date_cols_reviews] = order_reviews_df[date_cols_reviews].apply(pd.to_datetime)
```

```python
#To check wether all customer_id values in orders_df exist in customers_df.customer_id :

# orders_df['customer_id'].isin(customers_df['customer_id']) --> gives boolean (True if exists)

# ~orders_df['customer_id'].isin(customers_df['customer_id']) --> reverses the boolean series (True -> False)

(~orders_df['customer_id'].isin(customers_df['customer_id'])).sum()
```

```python
# To check wether all order id vlaues in order_items_df exist in orders_df.order_id:

(~order_items_df['order_id'].isin(orders_df['order_id'])).sum()
```

```python
# Does every product_id in order_items_df exist in products_df? :

(~order_items_df['product_id'].isin(products_df['product_id'])).sum()
```

```python
# Does every seller_id in order_items_df exist in sellers_df? :

(~order_items_df['seller_id'].isin(sellers_df['seller_id'])).sum()
```

```python
# Does every order_id in order_reviews_df exist in orders_df? :

(~order_reviews_df['order_id'].isin(orders_df['order_id'])).sum()
```

```python
# Merging orders_df and customers_df :

orders_customers_df = orders_df.merge(
    customers_df,
    on= "customer_id",
    how= "left"
)


orders_customers_df.shape
orders_customers_df['order_id'].nunique()
orders_customers_df["customer_id"].isna().sum()
```

```python
# Merging order_items_df to orders_customers_df :

orders_items_customers_df = orders_customers_df.merge(
    order_items_df,
    on= "order_id",
    how= "left"
)

print(orders_items_customers_df.shape)
print(orders_items_customers_df['order_id'].nunique())
print(orders_items_customers_df['order_id'].isna().sum())
```

```python
# Merging products_df to orders_items_customers_df : 

orders_items_customers_products_df = orders_items_customers_df.merge(
    products_df,
    on= "product_id",
    how= "left"
)

orders_items_customers_products_df.shape
```

```python
## Checking if one order has multiple reviews:
(order_reviews_df['order_id'].value_counts() > 1).sum()

## Checking if one order has multiple payments:
(order_payments_df['order_id'].value_counts() > 1).sum()

## Finding if orders with multiple payments are necessarily duplicate:
order_payments_df[order_payments_df['order_id'].duplicated(keep=False)].sort_values("order_id").head(10)
```

```python
## Getting only 1 total_payment record per 1 order:

order_level_payment_summary_df = (

    order_payments_df.groupby("order_id").agg(

        total_payment = ('payment_value', 'sum')
    )
) 

order_level_payment_summary_df = order_level_payment_summary_df.reset_index()
order_level_payment_summary_df
```

```python
#Getting average review for 1 order to get 1 total review per 1 order:

order_avg_reviews_df = (

    order_reviews_df.groupby('order_id').agg(

        avg_review_score = ('review_score', 'mean'),
        review_count =  ('review_score', 'count')
    ).reset_index()
)

order_avg_reviews_df
```

```python
## Merging avg_reviews_df to the previous df(all in one df):

df_main = orders_items_customers_products_df.merge(
    order_avg_reviews_df,
    on= "order_id",
    how= "left"
)

df_main.shape
```

```python
## Merging total_payments_df to the previous df(all in one df):

df_main_2 = df_main.merge(
    order_level_payment_summary_df,
    on= "order_id",
    how= "left"
)

df_main_2.shape
```

```python
## Merging sellers_df to get one Final Main Df (all in one df):

df_final = df_main_2.merge(
    sellers_df,
    on= "seller_id",
    how= "left"
)

df_final.shape
```

```python
## Let's verify how many unique orders we have: 

df_final['order_id'].nunique()
```

```python
#Let's verify how many unique products we have:

df_final['product_id'].nunique()
```

```python
df_final.info()

print(df_final.isna().sum())
```

```python
##Q1) Which product category generates the highest total revenue?

result_q1 = (

    df_final.groupby('product_category_name').agg(
        total_revenue = ('price', 'sum')
    )
    .sort_values(by='total_revenue',ascending=False).head(5)
)
result_q1
```

```python
##Q2) Which product category has the highest average item price?

result_q2 = (

    df_final.groupby('product_category_name').agg(

        avg_item_price = ('price', 'mean')
    )
    .sort_values(by='avg_item_price', ascending=False).head(5)
)

result_q2
```

```python
##Q3) Which product category has the highest number of items sold?

result_q3 = (

    df_final.groupby('product_category_name').agg(

        no_of_items_sold = ('order_id', 'count')
    )
    .sort_values(by= 'no_of_items_sold', ascending=False).head(5)
)

result_q3
```

```python
##Q4) Which product category has the highest total freight cost?

result_q4 = (

    df_final.groupby('product_category_name').agg(

        total_freight_cost = ('freight_value', 'sum')
    )
    .sort_values(by= 'total_freight_cost', ascending= False).head(5)
)

result_q4
```

```python
##Q5) For each product category, what is the average freight cost per item?

result_q5 = (

    df_final.groupby('product_category_name').agg(

        avg_freight_cost = ('freight_value', 'mean')
    )
    .sort_values(by='avg_freight_cost', ascending=False)
)

result_q5
```

```python
##Q6 Which product category has the highest average total cost per item, where total cost = price + freight_value?

df_final['total_cost'] = df_final['price'] + df_final['freight_value']

result_q6 = (

    df_final.groupby('product_category_name').agg(

        avg_total_cost_item = ('total_cost', 'mean')
    )
    .sort_values(by= 'avg_total_cost_item', ascending=False).head(5)
)

result_q6
```

```python
##Q7) Which customer state has the highest number of orders?



result_q7 = (

    df_final.groupby('customer_state').agg(

        no_of_orders = ('order_id', 'nunique')
    )
    .sort_values(by='no_of_orders', ascending=False).head()
)

result_q7
```

```python
##Q8) Which customer state generates the highest total revenue?

result_q8 = (

    df_final.groupby('customer_state').agg(

        total_revenue = ('price', 'sum')
    )
    .sort_values(by= 'total_revenue', ascending=False).head(5)
)
result_q8
```

```python
##Q9) Which customer state has the highest average order value (AOV)? Here, AOV = total revenue / number of unique orders.

result_q9 = (

    df_final.groupby('customer_state').agg(

        total_revenue = ('price', 'sum'),
        unique_orders = ('order_id', 'nunique')
    )
)

result_q9['AOV'] = result_q9['total_revenue'] / result_q9['unique_orders']

result_q9 = (
    result_q9.sort_values(by="AOV", ascending=False)
    .head(5)
)

result_q9
```

```python
##Q10) What percentage of customers have placed more than one order?

orders_per_customer = (

    df_final.groupby('customer_unique_id')['order_id']
    .nunique()
    .sort_values(ascending=False)
)

repeat_customers = orders_per_customer[orders_per_customer > 1]

percentage = len(repeat_customers) / len(orders_per_customer) * 100

percentage
```

```python
##Q11) Which customer state has the highest average order value among customers who have placed more than one order?

q11_series_1 = (

    df_final.groupby('customer_unique_id')['order_id']
    .nunique()
    .sort_values(ascending=False)
)

q11_repeat_cust = q11_series_1[q11_series_1 > 1]

df_repeat = df_final[
    df_final['customer_unique_id'].isin(q11_repeat_cust.index)   ## New learning 
]

result_q11 = (

    df_repeat.groupby('customer_state').agg(

        total_revenue = ('price', 'sum'),
        unique_orders = ('order_id', 'nunique')
    )
)

result_q11['AOV'] = result_q11['total_revenue'] / result_q11['unique_orders'] 

result_q11 = result_q11.sort_values(by="AOV", ascending=False).head(5)

result_q11
```

```python
##Q12) Which product categories have the highest average review score, and how many reviews does each category have?

q12_df_1 = order_reviews_df.merge(
    order_items_df,
    on= 'order_id',
    how= 'left'
)

q12_final_df = q12_df_1.merge(
    products_df,
    on="product_id",
    how='left'
)

result_q12 = (

    q12_final_df.groupby('product_category_name').agg(

        avg_review_score = ('review_score', 'mean'),
        review_count = ('review_score', 'count')
    )
    .sort_values(by='avg_review_score', ascending=False)
    .reset_index()
    
)

result_q12
```

```python
##Q13) Which product categories have both a high average review score and a sufficiently large number of reviews?

q13_df = result_q12[
    result_q12['review_count'] > 100
]

q13_df
```

```python
##Q14) Which product categories have the highest revenue but relatively low average review scores?


q14_df = (

    q12_final_df.groupby('product_category_name').agg(

        total_revenue = ('price', 'sum'),
        avg_review_score = ('review_score', 'mean'),
        reviews = ('review_score', 'count')
    )
    .sort_values(by='total_revenue', ascending=False)
    
)

result_q14_df = q14_df[(

    q14_df['total_revenue'] > 500000) &(
        q14_df['avg_review_score'] < 4
    )
]

result_q14_df
```

```python
##Q15) Which customer states have the highest percentage of orders delivered later than the estimated delivery date?


df_final['late_delivery'] = (

    df_final['order_delivered_customer_date'] > df_final['order_estimated_delivery_date']
)

q15_df = df_final[df_final['order_delivered_customer_date'].notna()]

result_q15 = (

    q15_df.groupby('customer_state')['late_delivery'].mean()* 100
    
)

result_q15 = (
    result_q15.reset_index().rename(columns={'late_delivery': 'late_del_percent'})
    .sort_values(by='late_del_percent', ascending=False)
    .head(5)
              )
result_q15
```

```python
#Q16)What percentage of orders were delivered late overall, and does the result change significantly when calculated at the order level instead of the item level?

q16_df = df_final[df_final['order_delivered_customer_date'].notna()]

overall_late_del_percent = (

    q16_df['late_delivery'].mean() * 100
)

print(f"Overall late delivery percentage : {overall_late_del_percent}")

order_level_df = (
    q16_df.groupby('order_id').agg(
        customer_state = ('customer_state', 'first'),
        late_delivery = ('late_delivery', 'first')
    )
    .reset_index()
)

result_q16 = (
    order_level_df['late_delivery'].mean() * 100
)

print(f"Order level late del percent: {result_q16}")

customer_state_late_del_perct = (

    order_level_df.groupby('customer_state')['late_delivery']
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .reset_index()
    .head(5)
)

difference = (

    result_q16 - overall_late_del_percent
)

print(f"Difference : {difference}")


customer_state_late_del_perct
```

```python
#Q17) Which customer states have the highest average delivery delay, considering only orders that were delivered late?


order_level_delivery_df = (
    df_final[df_final['order_delivered_customer_date'].notna()]           # Creating and order_level_delivery_df
    .groupby('order_id')
    .agg(
        customer_state=('customer_state', 'first'),
        late_delivery=('late_delivery', 'first'),
        delivered_date=('order_delivered_customer_date', 'first'),
        estimated_date=('order_estimated_delivery_date', 'first')
    )
    .reset_index()
)

order_level_delivery_df['delay_days'] = (
    order_level_delivery_df['delivered_date']                 # Calculating delay_days  
    - order_level_delivery_df['estimated_date']
).dt.total_seconds() / 86400


result_q17 = (
    order_level_delivery_df[                                   # Filtered only orders that were delivered late.
        order_level_delivery_df['late_delivery']             
    ]
    .groupby('customer_state')['delay_days']
    .mean()
    .sort_values(ascending=False)
    .reset_index(name='avg_delay_days')
    .head(5)
)

result_q17
```

```python
#Q18) Which customer states have both a high late-delivery rate and a high average delivery delay?

result_q18 = (
    order_level_delivery_df
    .groupby('customer_state')
    .agg(
        late_delivery_rate=('late_delivery', lambda x: x.mean() * 100),         # x represents all values in late_delivery column for a state 
        avg_delay_days=('delay_days', lambda x: x[x > 0].mean())                # x[x  > 0 ] filters orders that were late and x represents all values in delay_days
    )
    .reset_index()
    .sort_values(by=['late_delivery_rate', 'avg_delay_days'], ascending=[False,False])
)

result_q18 = result_q18[(                                         # Filtering df with reasonable thresholds 
    result_q18['late_delivery_rate'] > 10                       
) & (
    result_q18['avg_delay_days'] > 13
)]

result_q18
```

```python
#Q19) Which product categories have the highest cancellation/unavailable order rates?
      # 1 row = 1 order item

q19_df = df_final[
    ['order_id', 'product_category_name', 'order_status']
].copy()

q19_df['problem_order'] = q19_df['order_status'].isin(
    ['canceled', 'unavailable']
)


result_q19 = (
    q19_df.groupby('product_category_name').agg(
        orders = ('order_id', 'count'),
        problem_orders = ('problem_order', 'sum')
    )
    .reset_index()
)

result_q19['problem_orders_rate'] = result_q19['problem_orders'] / result_q19['orders'] * 100

result_q19 = (
    result_q19.sort_values(by='problem_orders_rate', ascending=False)
    .head(10)
)

result_q19
```

```python
#Q19B) Same Question But Unique Orders ? 
       # 1 row = 1 order + 1 category

q19b_df = df_final[
    ['order_id', 'product_category_name', 'order_status']
].copy()


q19b_df = q19b_df.drop_duplicates(                # the combination of these two columns defines what "duplicate" means for this operation.
    subset= ['order_id', 'product_category_name']     #A + electronics  ← first time → KEEP 
                                                      #A + electronics  ← seen before → REMOVE
)

q19b_df['problem_order'] = q19b_df['order_status'].isin(
    ['canceled', 'unavailable']
)

result_q19b = (
    q19b_df.groupby('product_category_name').agg(
        orders = ('order_id', 'nunique'),
        problem_orders = ('problem_order', 'sum')
    )
    .reset_index()
)

result_q19b['problem_order_rate'] = result_q19b['problem_orders'] / result_q19b['orders'] * 100

result_q19b = result_q19b.sort_values(by='problem_order_rate',ascending=False)
result_q19b
```

```python
#Q20) Which product categories have the highest average number of items per order?
      # 1 row = 1 unique order + 1 category 

items_per_order_category = (
    df_final.groupby(                                              # could have used .count too but,
        ['product_category_name', 'order_id']                      # .size() counts rows or elements and non nulls too.
    )
    .size()
    .reset_index(name="item_count")
)

result_q20 = (
    items_per_order_category.groupby('product_category_name').agg(             # groupby product category and then finding mean of item_count
        avg_items_per_order = ('item_count', 'mean')                           
    )
    .reset_index()
    .sort_values(by='avg_items_per_order',ascending=False)
    .head(10)
)

result_q20

```

```python
#Q21) Which product categories have the highest repeat-purchase rate?

q21_df = df_final[
    ['customer_unique_id', 'order_id', 'product_category_name']
].copy()

customer_category_orders = (

    q21_df.groupby(['customer_unique_id','product_category_name'])['order_id']
    .nunique()
    .reset_index(name='orders')
)

repeat_purchase_df = customer_category_orders[
    customer_category_orders['orders'] > 1
]

total_unique_customers_category = (

    customer_category_orders.groupby('product_category_name').agg(

        total_customers = ('customer_unique_id', 'nunique')
    )
    .reset_index()
)

repeat_customers_category = (

    repeat_purchase_df.groupby('product_category_name').agg(

        repeat_customers = ('customer_unique_id', 'nunique')
    )
    .reset_index()
)

q21_result_df = (total_unique_customers_category.merge(

    repeat_customers_category,
    on= 'product_category_name',
    how='left'
)
.fillna({'repeat_customers' : 0})
)

q21_result_df['repeat_purchase_rate'] = q21_result_df['repeat_customers'] / q21_result_df['total_customers'] * 100

q21_result_df = (
    q21_result_df
    .sort_values(by= 'repeat_purchase_rate', ascending=False)
)

q21_result_df
```

```python
#Q22) Which product categories generate high revenue but have a low repeat-purchase rate?

df_final_price_non_null = df_final[
    df_final['price'].notna()
].copy()


product_category_revenue = (

    df_final_price_non_null.groupby('product_category_name').agg(

        total_revenue = ('price', 'sum')
    )
    .sort_values(by= 'total_revenue', ascending=False)
    .reset_index()
)

revenue_by_category = (

    product_category_revenue.merge(
        q21_result_df[['product_category_name', 'repeat_purchase_rate']],
        on= 'product_category_name',
        how = 'inner'
    )
)

revenue_by_category = revenue_by_category[
    (
        revenue_by_category['total_revenue'] > 301660 ) &(
        revenue_by_category['repeat_purchase_rate'] < 4
    )
    .head(5)
]

revenue_by_category
```

```python
#Q23) Which product categories have high average order value but low repeat-purchase rates?

product_category_aov = (

    df_final_price_non_null.groupby('product_category_name').agg(

        total_revenue = ('price', 'sum'),
        unique_orders = ('order_id', 'nunique')
    )
    .reset_index()
)

product_category_aov['AOV'] = (
    product_category_aov['total_revenue'] / product_category_aov['unique_orders']
)

q23_result = ( product_category_aov[['product_category_name', 'AOV' ]].merge(
    q21_result_df[['product_category_name', 'repeat_purchase_rate']],
    on= 'product_category_name',
    how= 'left'
)
.sort_values(by= 'AOV', ascending=False)
)

q23_result = q23_result[(
    q23_result['AOV'] > 303
) & (
    q23_result['repeat_purchase_rate'] < 1.5
)
]

q23_result
```

```python
#Q24) Which product categories have the highest average product price but the lowest average review scores?

q24_df = q12_final_df[q12_final_df['price'].notna()]

category_avg_price_reviews = (

    q24_df.groupby('product_category_name').agg(

        average_price = ('price', 'mean'),
        average_review = ('review_score', 'mean')
    )
    .reset_index()
)

result_q24 = category_avg_price_reviews[(
    category_avg_price_reviews['average_price'] > 146) & (
        category_avg_price_reviews['average_review'] < 3.5
    )   
]
result_q24 = result_q24.sort_values(by= 'average_price', ascending=False)
result_q24
```

```python
#Q27) Do repeat customers generate more revenue per customer than one-time customers?

customer_unique_order_revenue = (
    df_final_price_non_null.groupby('customer_unique_id').agg(

        unique_orders = ('order_id', 'nunique'),
        revenue = ('price', 'sum')
    )
    .reset_index()
)

repeat_customers_df = customer_unique_order_revenue[
    customer_unique_order_revenue['unique_orders'] > 1
]

repeat_customers_revenue_per_customer = repeat_customers_df['revenue'].mean()

one_time_customers_df = customer_unique_order_revenue[
    customer_unique_order_revenue['unique_orders'] == 1
]

one_time_customers_revenue_per_customer = one_time_customers_df['revenue'].mean()

print(f"One-Time Customer Revenue Per Customer: {one_time_customers_revenue_per_customer}")
print(f"Repeat Customers Revenue Per Customer: {repeat_customers_revenue_per_customer}")

if repeat_customers_revenue_per_customer > one_time_customers_revenue_per_customer:
    print("Yes , Repeat Customers Generate More Revenue")
else:
    print("No")
```
